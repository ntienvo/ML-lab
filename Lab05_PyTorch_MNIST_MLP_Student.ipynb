### A2. Download and transform MNIST
MNIST contains 28 × 28 grayscale images of handwritten digits. `ToTensor()` converts images to tensors with pixel values in [0, 1].

```python
# TODO: Define a transform using transforms.ToTensor().
# TODO: Create MNIST training and test datasets in root="./data".
# Use download=True and pass the transform to both datasets.
# Remember: train=True for training data and train=False for test data.
# TODO: Print the number of examples in each dataset.
tranforms = transforms.ToTensor()
train_dataset = datasets.MNIST(root="./data", train=True, download=True, transform=tranforms)
test_dataset = datasets.MNIST(root="./data", train=False, download=True, transform=tranforms)
print("Number of training examples:", len(train_dataset))
print("Number of test examples:", len(test_dataset))
```
